# ML Practical 1 — Uber Fare Prediction
**Aim:** Predict Uber ride fare using Linear Regression and Random Forest Regression.

**Dataset required:** `uber.csv`  
Place `uber.csv` in the same folder as this notebook before running the data-loading cell.

**Tasks:** preprocess data, identify/remove obvious outliers, examine correlation, train both regression models, and compare R², RMSE, and MAE.

## 1. Import libraries
Run this cell first. If a package is missing, install it in a separate cell with `%pip install pandas numpy matplotlib seaborn scikit-learn`.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from math import radians, sin, cos, asin, sqrt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

RANDOM_STATE = 42
```

## 2. Load and inspect the dataset

```python
df = pd.read_csv("uber.csv")
print("Initial shape:", df.shape)
display(df.head())
print("\nMissing values by column:")
display(df.isnull().sum())
```

## 3. Pre-process the data
Remove missing rows, convert the pickup timestamp, and extract date/time features.

```python
df = df.dropna().copy()

if "pickup_datetime" in df.columns:
    df["pickup_datetime"] = pd.to_datetime(df["pickup_datetime"], errors="coerce", utc=True)
    df = df.dropna(subset=["pickup_datetime"])
    df["hour"] = df["pickup_datetime"].dt.hour
    df["day"] = df["pickup_datetime"].dt.day
    df["month"] = df["pickup_datetime"].dt.month
    df["year"] = df["pickup_datetime"].dt.year
    df["day_of_week"] = df["pickup_datetime"].dt.dayofweek

df = df.drop(columns=["pickup_datetime", "key"], errors="ignore")
print("Shape after preprocessing:", df.shape)
display(df.head())
```

## 4. Identify and remove obvious outliers
This follows the practical's basic approach: remove non-positive/extreme fares and invalid latitude/longitude values.

```python
required_cols = [
    "fare_amount", "pickup_longitude", "pickup_latitude",
    "dropoff_longitude", "dropoff_latitude"
]
missing_cols = [c for c in required_cols if c not in df.columns]
if missing_cols:
    raise ValueError(f"Dataset is missing required columns: {missing_cols}")

print("Fare summary before filtering:")
display(df["fare_amount"].describe())

plt.figure(figsize=(8, 4))
sns.boxplot(x=df["fare_amount"])
plt.title("Fare Amount — Before Outlier Filtering")
plt.show()

df = df[
    (df["fare_amount"] > 0) & (df["fare_amount"] < 100) &
    df["pickup_latitude"].between(-90, 90) &
    df["dropoff_latitude"].between(-90, 90) &
    df["pickup_longitude"].between(-180, 180) &
    df["dropoff_longitude"].between(-180, 180)
].copy()

print("Shape after outlier filtering:", df.shape)
```

## 5. Calculate trip distance using the Haversine formula

```python
def haversine_km(lon1, lat1, lon2, lat2):
    # Convert decimal degrees to radians
    lon1, lat1, lon2, lat2 = map(radians, [lon1, lat1, lon2, lat2])
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    return 6371 * 2 * asin(sqrt(a))

df["distance_km"] = df.apply(
    lambda row: haversine_km(
        row["pickup_longitude"], row["pickup_latitude"],
        row["dropoff_longitude"], row["dropoff_latitude"]
    ),
    axis=1
)

# Remove impossible/extreme computed distances and trips with no movement
df = df[df["distance_km"].between(0, 200)].copy()
print("Shape after distance calculation:", df.shape)
display(df[["distance_km", "fare_amount"]].head())
```

## 6. Check correlation

```python
numeric_df = df.select_dtypes(include=np.number)
plt.figure(figsize=(12, 8))
sns.heatmap(numeric_df.corr(), cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

display(numeric_df.corr()["fare_amount"].sort_values(ascending=False))
```

## 7. Select features and split the data
The target is `fare_amount`. Identifier columns are not used as features.

```python
feature_cols = [
    "pickup_longitude", "pickup_latitude",
    "dropoff_longitude", "dropoff_latitude",
    "distance_km"
]
for col in ["hour", "day", "month", "year", "day_of_week", "passenger_count"]:
    if col in df.columns:
        feature_cols.append(col)

X = df[feature_cols].replace([np.inf, -np.inf], np.nan)
y = df["fare_amount"]
valid = X.notna().all(axis=1) & y.notna()
X, y = X.loc[valid], y.loc[valid]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE
)
print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
print("Features:", feature_cols)
```

## 8. Train Linear Regression

```python
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)
y_pred_lr = linear_model.predict(X_test)
print("Linear Regression model trained.")
```

## 9. Train Random Forest Regression
This can take a little longer on a large dataset.

```python
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=RANDOM_STATE,
    n_jobs=-1
)
rf_model.fit(X_train, y_train)
y_pred_rf = rf_model.predict(X_test)
print("Random Forest model trained.")
```

## 10. Evaluate and compare models
RMSE is the square root of mean squared error.

```python
def regression_metrics(name, y_true, y_pred):
    return {
        "Model": name,
        "R2 Score": r2_score(y_true, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
        "MAE": mean_absolute_error(y_true, y_pred)
    }

results = pd.DataFrame([
    regression_metrics("Linear Regression", y_test, y_pred_lr),
    regression_metrics("Random Forest Regression", y_test, y_pred_rf)
])
display(results)
```

## 11. Actual vs predicted fare

```python
plt.figure(figsize=(7, 6))
plt.scatter(y_test, y_pred_lr, alpha=0.35, label="Linear Regression")
plt.scatter(y_test, y_pred_rf, alpha=0.35, label="Random Forest")
low = min(y_test.min(), y_pred_lr.min(), y_pred_rf.min())
high = max(y_test.max(), y_pred_lr.max(), y_pred_rf.max())
plt.plot([low, high], [low, high], linestyle="--")
plt.xlabel("Actual Fare")
plt.ylabel("Predicted Fare")
plt.title("Actual vs Predicted Uber Fare")
plt.legend()
plt.tight_layout()
plt.show()
```

## Result
The notebook trains Linear Regression and Random Forest Regression and compares their R², RMSE, and MAE scores. The model with higher R² and lower RMSE/MAE performs better on this test split.

**Note:** Exact scores depend on the `uber.csv` dataset and the rows remaining after preprocessing.
