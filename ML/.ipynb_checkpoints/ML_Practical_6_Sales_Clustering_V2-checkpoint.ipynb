```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.cluster import KMeans, AgglomerativeClustering
```

```python
df = pd.read_csv("sales_data_sample.csv", encoding="latin1")
num = df.select_dtypes(include=np.number).copy()
num = num.drop(columns=[c for c in ["ORDERNUMBER", "ORDERLINENUMBER", "YEAR_ID", "QTR_ID", "MONTH_ID", "DAY_ID"] if c in num.columns], errors="ignore")
num = num.loc[:, num.nunique(dropna=True) > 1]
X = SimpleImputer(strategy="median").fit_transform(num)
X = StandardScaler().fit_transform(X)
print("Rows:", X.shape[0], "Features:", X.shape[1])
```

```python
scores = []
ks = range(1, 11)
for k in ks:
    model = KMeans(n_clusters=k, random_state=7, n_init=10)
    model.fit(X)
    scores.append(model.inertia_)
plt.plot(list(ks), scores, marker="o")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.xticks(list(ks))
plt.show()
```

```python
k = 4
km = KMeans(n_clusters=k, random_state=7, n_init=10)
labels = km.fit_predict(X)
print(pd.Series(labels).value_counts().sort_index())
if X.shape[1] >= 2:
    plt.scatter(X[:, 0], X[:, 1], c=labels)
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    plt.show()
```

```python
hierarchical = AgglomerativeClustering(n_clusters=k)
hier_labels = hierarchical.fit_predict(X)
print(pd.Series(hier_labels).value_counts().sort_index())
```
