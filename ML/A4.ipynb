```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return (x + 3) ** 2

def derivative(x):
    return 2 * (x + 3)

x = 2.0
learning_rate = 0.1
iterations = 100
history = [x]

for i in range(iterations):
    x = x - learning_rate * derivative(x)
    history.append(x)

curve_x = np.linspace(-7, 3, 300)
curve_y = f(curve_x)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(curve_x, curve_y, label="y = (x + 3)^2", color="blue")
plt.scatter(history, [f(v) for v in history], color="red", zorder=5, label="Gradient Steps")
plt.title("Gradient Descent Optimization")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```
