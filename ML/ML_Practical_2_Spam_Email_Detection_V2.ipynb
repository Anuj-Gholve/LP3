```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
df = pd.read_csv("emails.csv", encoding="ISO-8859-1")
df = df.drop(columns=["Email No."], errors="ignore")
df["Prediction"] = pd.to_numeric(df["Prediction"], errors="coerce")
df = df.dropna(subset=["Prediction"])
X = df.drop(columns="Prediction").apply(pd.to_numeric, errors="coerce")
X = X.replace([np.inf, -np.inf], np.nan)
X = X.fillna(X.median()).fillna(0)
y = df["Prediction"].astype(int)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=7, stratify=y)
```

```python
models = {"KNN": make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=7)), "SVM": make_pipeline(StandardScaler(), SVC(kernel="rbf"))}
results = []
for name, model in models.items():
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    print(name)
    print("Accuracy:", round(accuracy_score(y_test, pred), 4))
    print(confusion_matrix(y_test, pred))
    print(classification_report(y_test, pred, zero_division=0))
    results.append([name, accuracy_score(y_test, pred)])
display(pd.DataFrame(results, columns=["Model", "Accuracy"]).sort_values("Accuracy", ascending=False))
```
