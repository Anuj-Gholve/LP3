```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
f = lambda x: (x + 3) ** 2
gradient = lambda x: 2 * (x + 3)
x = 2.0
rate = 0.1
values = [x]
for step in range(100):
    updated = x - rate * gradient(x)
    values.append(updated)
    print(step + 1, round(updated, 6), round(f(updated), 6))
    if abs(updated - x) < 1e-6:
        x = updated
        break
    x = updated
print("Minimum x:", x)
print("Minimum y:", f(x))
```

```python
curve = np.linspace(-8, 3, 300)
plt.plot(curve, f(curve))
plt.scatter(values, [f(v) for v in values])
plt.xlabel("x")
plt.ylabel("f(x)")
plt.show()
```
